```python
print("\nDataset Information:")
df.info()
```

```python
print("Passenger names")
print(df["Name"].head())

print("Passenger by Gender:")
print(df["Pclass"].value_counts())

print("Average Age:")
print(df["Age"].mean())


print("Highest Fare:")
print(df["Fare"].max())

print("Average Fare:")
print(df["Fare"].mean())
```

```python
df1 = df.drop(['Cabin', 'Embarked'], axis=1, errors='ignore')
print(df1)
```

```python
print("Passengers older than 60")
print("Passengers Older Than 60")
print(df[df["Age"]>60][["Name", "Age", "Sex"]])
```

```python
#Display Passengers who survived

print("Passengers who survived:")
print(df[df["Survived"]==1][["Name", "Age", "Sex"]].head())
```

```python
#Count Survivors and non-survivors

print("Survival Counts")
print(df["Survived"].value_counts())
```

```python
#Calculate Survival Rate By Gender

print("Survival Rate By Gender:")
print(df.groupby("Sex")["Survived"].mean()*100)
```

```python
print("10 Highest Fares")
print(
    df.sort_values("Fare", ascending=False)
      [["Name","Fare"]].head(10)
)
```
